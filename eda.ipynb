## Conclusion1: the evolutions of price of extra virgen olive oil presents a tendency

```python
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf
plot_acf(extra_virgen['price_100kg'], lags=100)
```

```python
plot_pacf(extra_virgen['price_100kg'])
```

### the partial correlation decrease drastically after the 2  rd lag, and the rest of the point are inside the zone of confidence, we can conclude that AR autoregression will suit to model with p=2 

### As the time serie is not a stationary and presents a tendency, so we will apply a diferenciation and d=1

### We will be using q=0 for Movinge agerage 

```python
from statsmodels.tsa.arima.model import ARIMA
model = ARIMA(extra_virgen['price_100kg'], order = (2,1,0)).fit()
```

```python
import seaborn as sns
sns.lineplot(extra_virgen['price_100kg'])
sns.lineplot(model.predict())
```

```python
bic = None

for i in 0,1,2,3,4:
    # calcula AR(i)
    model_AR = ARIMA(extra_virgen['price_100kg'], order=(i,0,0)).fit()
    if bic is None or model_AR.bic < bic:
        bic = model_AR.bic
        final = model_AR
        result = ("AR",i)

    # calcula MA(i)
    model_MA = ARIMA(extra_virgen['price_100kg'], order=(0,0,i)).fit()
    if bic is None or model_MA.bic < bic:
        bic = model_MA.bic
        final = model_MA
        result = ("MA",i)

print(result)
```

```python
from sklearn.model_selection import train_test_split
```

```python
train = extra_virgen.loc[:'2022-12-31', 'price_100kg']
test = extra_virgen.loc['2023-01-01':, 'price_100kg']
```

```python
model = ARIMA(train, order = (3,1,0)).fit()
```

### Forecasting

```python
forecast = model.forecast(steps=len(test))
plt.figure(figsize=(10,5))

plt.plot(train.index, train, label="Train")
plt.plot(test.index, test, label="Observed")
plt.plot(test.index, forecast, label="Forecast")

plt.legend()
plt.show()
```
